**Implementation of Stack**

```python
stack = []
top = -1  # Empty stack starts at -1

# PUSH (Increment top)
top += 1
stack.append(10)

top += 1
stack.append(20)

top += 1
stack.append(30)

top += 1
stack.append(40)


print("Stack:", stack)  
print("Top index:", top) 

# POP (Decrement top)
stack.pop()
top -= 1

print("Stack:", stack) 
print("Top index:", top)
```

**Implementation Of Queue**

```python
Queue = []
front = -1
rear = -1

# ADD to Queue (Enqueue)

def enqueue(value):
    global front, rear
    if front == -1 and rear == -1:
        front += 1         
        rear += 1          
        Queue.append(value)
    else:
        rear += 1          
        Queue.append(value)

# REMOVE from Queue (Dequeue)
def dequeue():
    global front, rear
    if front == -1 or front > rear:
        print("Queue is empty!")
    else:
        removed = Queue[front]  
        front += 1              
        print("Removed:", removed)

# --- Testing your values ---
enqueue(10)   
enqueue(20) 
enqueue(30)   

dequeue()     # Removes 10 -> front becomes 1 (now pointing at 20)

# View active queue from front to rear
print("Current Queue:", Queue[front : rear + 1])
```
