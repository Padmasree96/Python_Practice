Count frequency of each element in an array

```python
arr = [1, 2, 2, 3, 3, 3, 4, 4]

freq = {}

for i in arr:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

print(freq)
```

Find all the repeating elements

```python
arr = [1, 2, 2, 3, 3, 3, 4, 4]

freq = {}

for i in arr:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

for i in freq:
    if freq[i] > 1:
        print(i)
```

Find all the non repeating elements

```python
arr = [1, 2, 2,3, 4, 4]

freq = {}

for i in arr:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

for i in freq:
    if freq[i] == 1:
        print(i)
```

Find distinct elements

```python
arr = [1, 2, 2, 3, 3, 3, 4, 5]

result = []

for i in arr:
    if i not in result:
        result.append(i)

print(result)
```

First Repeating Element

```python
arr = [10, 20, 30, 20, 40, 10]

for i in range(len(arr)):
    for j in range(i + 1, len(arr)):
        if arr[i] == arr[j]:
            print(arr[i])
            break
    else:
        continue
    break
```

Non-Repeating Element

```python
arr = [10, 20, 30, 20, 40, 10]

for i in range(len(arr)):
    count = 0

    for j in range(len(arr)):
        if arr[i] == arr[j]:
            count += 1

    if count == 1:
        print(arr[i])
        break
```
