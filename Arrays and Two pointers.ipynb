Find Missing Number

```python
arr = [1, 2, 3, 5, 6]

for i in range(1, 7):
    if i not in arr:
        print(i)
```

Reverse an array

```python
arr = [10, 20, 30, 40]

i = 0
j = len(arr)-1

while i < j:
    arr[i], arr[j] = arr[j], arr[i]
    i += 1
    j -= 1

print(arr)
```

Remove Duplicates from Sorted Array

```python
nums = [1, 1, 2, 2, 3,4,4]

i = 0

for j in range(1, len(nums)):
    if nums[j] != nums[i]:
        i += 1
        nums[i] = nums[j]

print(i + 1)
print(nums[:i+1])
```

Move Zeroes

```python
nums = [0, 1, 0, 3, 12]



i = 0

for j in range(len(nums)):
    if nums[j] != 0:
        nums[i], nums[j] = nums[j], nums[i]
        i += 1

print(nums)
```
